```python
# Problema 1: Clasificador de temperatura

# Recorremos los números desde -30 hasta 50
for temperatura in range(-30, 51):

    # Si es menor que 0, se considera muy frío
    if temperatura < 0:
        print(temperatura, "Muy frío")

    # Si está entre 0 y 24, se considera templado
    elif temperatura <= 24:
        print(temperatura, "Templado")

    # Si es 25 o más, se considera caluroso
    else:
        print(temperatura, "Caluroso")
```

```python
# Problema 2: Números especiales del robot

# Recorremos los números desde 1 hasta 100
for numero in range(1, 101):

    # Comprobamos si el número es divisible por 5
    if numero % 5 == 0:

        # Si también es divisible por 10, imprime "Bip Bip"
        if numero % 10 == 0:
            print(numero, "Bip Bip")

        # Si solo es divisible por 5, imprime "Bip"
        else:
            print(numero, "Bip")
```

```python
# Problema 3: Detector de puntaje gamer

# Pedimos al usuario su primer puntaje
primer_puntaje = int(input("Ingresa primer puntaje: "))

# Pedimos al usuario su segundo puntaje
segundo_puntaje = int(input("Ingresa segundo puntaje: "))

# Sumamos ambos puntajes
resultado = primer_puntaje + segundo_puntaje

# Clasificamos al jugador según el resultado
if resultado < 50:
    print("Resultado: Nivel principiante")

elif resultado < 80:
    print("Resultado: Nivel intermedio")

else:
    print("Resultado: Nivel pro")
```

```python
# Problema 4: Calculador de perfil musical

# Pedimos la edad del usuario
edad = int(input("Ingresa tu edad: "))

# Preguntamos si le gusta la música
le_gusta_musica = input("¿Te gusta la música? (si/no): ")

# Preguntamos si le gusta bailar
le_gusta_bailar = input("¿Te gusta bailar? (si/no): ")

# Convertimos las respuestas a valores True o False
le_gusta_musica = le_gusta_musica == "si"
le_gusta_bailar = le_gusta_bailar == "si"

# Evaluamos el perfil según las condiciones
if edad >= 18 and le_gusta_musica and le_gusta_bailar:
    print("Eres una estrella de la fiesta")

elif le_gusta_musica and not le_gusta_bailar:
    print("Te gusta la música, pero no tanto bailar")

else:
    print("Tienes tu propio estilo")
```
